## Question 1(a) — Find the separating hyperplane by inspection

In two dimensions, a hyperplane is simply a **line**.

Look at the $x_1$ coordinates.

For every blue point,

$$
x_1\le0.
$$

For every red point,

$$
x_1\ge1.
$$

Therefore there is an empty gap between

$$
x_1=0
$$

and

$$
x_1=1.
$$

Any vertical line

$$
x_1=c,
\qquad
0<c<1,
$$

separates the two classes.

The most natural choice is the midpoint:

$$
\boxed{x_1=\frac12}.
$$

Write the boundary in standard perceptron form:

$$
w_1x_1+w_2x_2+b=0.
$$

Multiplying

$$
x_1-\frac12=0
$$

by 2 gives

$$
2x_1-1=0.
$$

Therefore one convenient representation is

$$
\boxed{
\mathbf{w}
=
\begin{bmatrix}
2\\
0
\end{bmatrix},
\qquad
b=-1
}
$$

because

$$
\mathbf{w}^T\mathbf{x}+b
=
2x_1-1.
$$

### Check the two sides

For the blue class,

$$
x_1\le0
\Longrightarrow
2x_1-1<0.
$$

For the red class,

$$
x_1\ge1
\Longrightarrow
2x_1-1>0.
$$

So the line separates every point correctly.

### Answer

$$
\boxed{
x_1=\frac12
}
$$

or equivalently

$$
\boxed{
2x_1-1=0.
}
$$

```python
# Plot the by-inspection separating line x1 = 0.5.

plt.figure(figsize=(7, 6))
plt.scatter(blue[:, 0], blue[:, 1], marker="o", s=90, label="Blue class")
plt.scatter(red[:, 0], red[:, 1], marker="x", s=90, label="Red class")

plt.axvline(0.5, linewidth=2, label="$x_1=0.5$")

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.title("Question 1(a) — Separating Line by Inspection")
plt.xlim(-1.5, 2.5)
plt.ylim(-1.5, 2.5)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

# Question 1(b) — Perceptron Algorithm

The professor asks us to initialize the perceptron using the line

$$
x_2=x_1.
$$

Rewrite it as

$$
x_1-x_2=0.
$$

Therefore a convenient initial perceptron is

$$
\boxed{
\mathbf{w}^{(0)}
=
\begin{bmatrix}
1\\
-1
\end{bmatrix},
\qquad
b^{(0)}=0.
}
$$

The decision score is

$$
a
=
\mathbf{w}^T\mathbf{x}+b.
$$

We use the class coding

$$
y=
\begin{cases}
+1,&\text{red},\\
-1,&\text{blue}.
\end{cases}
$$

For a training point $(\mathbf{x},y)$, the point is correctly classified when

$$
y(\mathbf{w}^T\mathbf{x}+b)>0.
$$

If

$$
y(\mathbf{w}^T\mathbf{x}+b)\le0,
$$

we update the perceptron.

With learning rate

$$
\eta=1,
$$

the standard update is

$$
\boxed{
\mathbf{w}
\leftarrow
\mathbf{w}+\eta y\mathbf{x}
}
$$

and

$$
\boxed{
b
\leftarrow
b+\eta y.
}
$$

---

## Deterministic training convention used here

The homework does not specify a unique data order or learning rate.

We therefore use the standard choices:

- learning rate $\eta=1$;
- blue label $-1$;
- red label $+1$;
- a point exactly on the current boundary is treated as requiring an update;
- process the plotted points in this order:

$$
(-1,0),
(-1,-1),
(0,-1),
(1,1),
(1,2),
(2,1),
(2,2).
$$

A linearly separable dataset can have many valid perceptron solutions, so the exact final weights can depend on the order. What matters is that the resulting line correctly separates all training points.

```python
# Question 1(b) — Perceptron training

X = np.array([
    [-1.0,  0.0],
    [-1.0, -1.0],
    [ 0.0, -1.0],
    [ 1.0,  1.0],
    [ 1.0,  2.0],
    [ 2.0,  1.0],
    [ 2.0,  2.0]
])

y = np.array([-1, -1, -1, 1, 1, 1, 1])

# x2 = x1  <=>  x1 - x2 = 0
w = np.array([1.0, -1.0])
b = 0.0
eta = 1.0

history = []

for epoch in range(100):
    mistakes = 0

    for i in range(len(X)):
        score = np.dot(w, X[i]) + b

        if y[i] * score <= 0:
            old_w = w.copy()
            old_b = b

            w = w + eta * y[i] * X[i]
            b = b + eta * y[i]

            history.append({
                "epoch": epoch + 1,
                "point": X[i].copy(),
                "label": int(y[i]),
                "old_w": old_w,
                "old_b": old_b,
                "old_score": score,
                "new_w": w.copy(),
                "new_b": b
            })

            mistakes += 1

    if mistakes == 0:
        break

print("Converged after", epoch + 1, "epoch(s)")
print("Final w =", w)
print("Final b =", b)
print("Number of updates =", len(history))

for j, h in enumerate(history, start=1):
    print(f"\nUpdate {j}")
    print(" point =", h["point"], "label =", h["label"])
    print(" old w =", h["old_w"], "old b =", h["old_b"])
    print(" old score =", h["old_score"])
    print(" new w =", h["new_w"], "new b =", h["new_b"])
```

## Manual explanation of the perceptron update

Start with

$$
\mathbf{w}^{(0)}
=
\begin{bmatrix}
1\\
-1
\end{bmatrix},
\qquad
b^{(0)}=0.
$$

The initial line is

$$
x_1-x_2=0,
$$

or

$$
x_2=x_1.
$$

### First blue point: $(-1,0)$

The score is

$$
a
=
(1)(-1)+(-1)(0)+0
=
-1.
$$

Its label is

$$
y=-1.
$$

So

$$
ya=(-1)(-1)=1>0.
$$

It is correctly classified.

No update.

---

### Second blue point: $(-1,-1)$

The score is

$$
a
=
(1)(-1)+(-1)(-1)+0
=
0.
$$

The point lies exactly on the current decision boundary.

Because

$$
ya=0,
$$

we perform an update.

The label is

$$
y=-1.
$$

The new weight vector is

$$
\mathbf{w}^{(1)}
=
\mathbf{w}^{(0)}
+
y\mathbf{x}.
$$

Therefore,

$$
\mathbf{w}^{(1)}
=
\begin{bmatrix}
1\\
-1
\end{bmatrix}
+
(-1)
\begin{bmatrix}
-1\\
-1
\end{bmatrix}.
$$

Hence,

$$
\mathbf{w}^{(1)}
=
\begin{bmatrix}
1\\
-1
\end{bmatrix}
+
\begin{bmatrix}
1\\
1
\end{bmatrix}
=
\begin{bmatrix}
2\\
0
\end{bmatrix}.
$$

For the bias,

$$
b^{(1)}
=
0+(-1)
=
-1.
$$

So after the update,

$$
\boxed{
\mathbf{w}
=
\begin{bmatrix}
2\\
0
\end{bmatrix},
\qquad
b=-1.
}
$$

The new boundary is

$$
2x_1-1=0.
$$

Thus,

$$
\boxed{
x_1=\frac12.
}
$$

This is exactly the natural separator identified in Question 1(a).

All remaining points are correctly classified by this boundary. A second pass produces zero errors, so the perceptron has converged.

### Final perceptron answer

$$
\boxed{
\mathbf{w}
=
\begin{bmatrix}
2\\
0
\end{bmatrix},
\qquad
b=-1
}
$$

and therefore

$$
\boxed{
2x_1-1=0
}
$$

or

$$
\boxed{
x_1=0.5.
}
$$

```python
# Verify the final classification of every point.

scores = X @ w + b
predictions = np.where(scores > 0, 1, -1)

print("Point        Label   Score   Prediction   Correct?")
print("-" * 56)

for point, target, score, pred in zip(X, y, scores, predictions):
    print(
        f"{str(point):14s} {target:5d} "
        f"{score:7.2f} {pred:11d} {str(pred == target):>10s}"
    )

assert np.array_equal(predictions, y)
print("\nAll training points are correctly separated.")
```

```python
# Plot the initial line and the final perceptron line.

xx = np.linspace(-1.5, 2.5, 300)

plt.figure(figsize=(8, 6))
plt.scatter(blue[:, 0], blue[:, 1], marker="o", s=90, label="Blue class (-1)")
plt.scatter(red[:, 0], red[:, 1], marker="x", s=90, label="Red class (+1)")

# Initial line x2 = x1
plt.plot(xx, xx, linestyle="--", linewidth=2, label="Initial: $x_2=x_1$")

# Final line x1 = 0.5
plt.axvline(0.5, linewidth=2, label="Final: $x_1=0.5$")

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.title("Question 1(b) — Initial and Learned Perceptron Boundaries")
plt.xlim(-1.5, 2.5)
plt.ylim(-1.5, 2.5)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

# Question 3 — Perceptron for $x_1+x_2-2=0$

The decision boundary is

$$
x_1+x_2-2=0.
$$

Compare this with

$$
w_1x_1+w_2x_2+b=0.
$$

Therefore,

$$
\boxed{
w_1=1,\qquad
w_2=1,\qquad
b=-2.
}
$$

The perceptron's net input is

$$
a
=
x_1+x_2-2.
$$

For a binary step activation, we can define

$$
\hat y
=
\begin{cases}
1,&a\ge0,\\
0,&a<0.
\end{cases}
$$

The exact labels $0/1$ or $-1/+1$ are conventions; what matters is the sign of the net input relative to the separating line.

## Question 3(a) — Block diagram

The perceptron performs these operations:

$$
x_1
\xrightarrow{\times 1}
$$

and

$$
x_2
\xrightarrow{\times 1}.
$$

The bias contributes

$$
1\xrightarrow{\times(-2)}.
$$

These terms are added:

$$
a
=
1x_1+1x_2-2.
$$

Then a step activation produces the class:

$$
\boxed{
(x_1,x_2)
\rightarrow
\text{weighted sum}
\rightarrow
\text{step activation}
\rightarrow
\text{class}.
}
$$

```python
# Question 3(a) — Draw the perceptron block diagram with Matplotlib.

fig, ax = plt.subplots(figsize=(10, 5))
ax.set_xlim(0, 10)
ax.set_ylim(0, 6)
ax.axis("off")

# Input nodes
ax.scatter([1, 1, 1], [4.5, 3.0, 1.5], s=900)
ax.text(1, 4.5, "$x_1$", ha="center", va="center", fontsize=14)
ax.text(1, 3.0, "$x_2$", ha="center", va="center", fontsize=14)
ax.text(1, 1.5, "$1$", ha="center", va="center", fontsize=14)

# Summation node
ax.scatter([5], [3], s=1200)
ax.text(5, 3, "$\Sigma$", ha="center", va="center", fontsize=20)

# Activation/output
ax.scatter([8], [3], s=1300)
ax.text(8, 3, "Step", ha="center", va="center", fontsize=13)

ax.text(9.4, 3, "Class", ha="center", va="center", fontsize=14)

# Arrows
def arrow(x1, y1, x2, y2, label):
    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        arrowprops=dict(arrowstyle="->", linewidth=1.8)
    )
    ax.text((x1+x2)/2, (y1+y2)/2 + 0.25, label, ha="center", fontsize=12)

arrow(1.5, 4.5, 4.4, 3.25, "$w_1=1$")
arrow(1.5, 3.0, 4.4, 3.0, "$w_2=1$")
arrow(1.5, 1.5, 4.4, 2.75, "$b=-2$")
arrow(5.6, 3.0, 7.4, 3.0, "$a=x_1+x_2-2$")
arrow(8.6, 3.0, 9.0, 3.0, "")

ax.set_title("Question 3(a) — Perceptron Block Diagram", fontsize=15)
plt.show()
```

# Question 4 — Why Use Deep Learning for Classification?

Deep learning is useful for classification because many real classification problems involve relationships that are too complex to represent with a single linear decision boundary.

A single perceptron creates a boundary of the form

$$
\mathbf{w}^T\mathbf{x}+b=0.
$$

That is a linear boundary.

However, real-world data such as images, speech, text, medical signals, and cybersecurity telemetry often require **nonlinear** decision functions.

A deep neural network addresses this by composing many layers:

$$
\boxed{
\text{input}
\rightarrow
\text{features}
\rightarrow
\text{higher-level features}
\rightarrow
\text{class}
}
$$

## Major reasons deep learning is useful

### 1. It can model nonlinear decision boundaries

Deep networks combine linear transformations with nonlinear activation functions.

This allows the final classifier to represent complex class boundaries.

---

### 2. It can learn useful features automatically

For an image, a deep network may progressively learn representations related to:

$$
\text{edges}
\rightarrow
\text{textures}
\rightarrow
\text{parts}
\rightarrow
\text{objects}.
$$

This reduces the need to manually design every feature.

---

### 3. It handles high-dimensional data

A small image can contain thousands of pixel values.

Modern neural networks can learn directly from these high-dimensional representations.

---

### 4. It supports multiclass classification

A network can produce scores for multiple classes.

For example,

$$
\text{image}
\rightarrow
\{\text{cat},\text{dog},\text{car},\text{person},\ldots\}.
$$

---

### 5. It supports end-to-end learning

Instead of separately designing features and then building a classifier, a deep network can learn both the representation and the classification rule from the training data.

---

## Important limitation

Deep learning is **not automatically the best method for every classification problem**.

It generally benefits from:

- sufficient training data;
- suitable architecture;
- computational resources;
- careful training and validation.

For a simple linearly separable dataset such as Question 1, a single perceptron is enough.

### Final answer

$$
\boxed{
\text{Deep learning is useful when classification requires complex, nonlinear, high-dimensional feature learning.}
}
$$

# Question 5 — Is Question 1 a Learning Problem?

## Short answer

$$
\boxed{\text{Yes — Question 1(b) is a supervised learning problem.}}
$$

## Why?

A learning problem contains:

1. examples;
2. target information;
3. parameters that are adjusted from data.

Question 1 provides labeled examples:

$$
(\mathbf{x}_i,y_i).
$$

The point coordinates are the inputs, and the red/blue classes are the target labels.

The perceptron starts from an initial decision boundary,

$$
x_2=x_1,
$$

which corresponds to initial parameters

$$
\mathbf{w}^{(0)}
=
\begin{bmatrix}
1\\
-1
\end{bmatrix},
\qquad
b^{(0)}=0.
$$

Whenever the model makes an error, it changes its parameters:

$$
\mathbf{w}
\leftarrow
\mathbf{w}+\eta y_i\mathbf{x}_i,
$$

$$
b
\leftarrow
b+\eta y_i.
$$

The final boundary is therefore determined by the training examples.

That is exactly the idea of supervised learning.

---

## Important distinction between 1(a) and 1(b)

Question 1(a) asks us to find a separating line **by inspection**.

That part is a human analytical solution rather than an automated learning procedure.

Question 1(b), however, explicitly uses the **perceptron training algorithm**, so it is a genuine learning process.

### Final answer

$$
\boxed{
\text{Question 1(b) is supervised learning because the perceptron learns its decision boundary from labeled examples.}
}
$$

# Question 6 — Is Question 1 Equivalent to Real-World Image/Object Recognition?

## Short answer

Question 1 and real-world image recognition are based on the **same general supervised-classification idea**, but they are **not equivalent in difficulty or model complexity**.

---

## How they are similar

In Question 1, each example is represented by a feature vector

$$
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}
$$

and is associated with a class label.

The model learns a function

$$
f(\mathbf{x})
\rightarrow
\text{class}.
$$

Image recognition has the same abstract structure.

An image is converted into numerical inputs, and the model predicts a class:

$$
\text{image data}
\rightarrow
\text{model}
\rightarrow
\text{object class}.
$$

So both are supervised classification problems.

---

## How they are different

### 1. Dimensionality

Question 1 uses only two features:

$$
x_1,\;x_2.
$$

An image can contain thousands or millions of pixel/channel values.

For example, an RGB image of size

$$
224\times224
$$

contains

$$
224\times224\times3
=
150{,}528
$$

raw numerical values.

---

### 2. Linear separability

Question 1 is linearly separable.

A single straight line separates the classes.

Real images are normally not separable directly by one simple hyperplane in raw pixel space.

---

### 3. Feature extraction

In Question 1, the two useful coordinates are already supplied.

In image recognition, the model often must learn useful features such as:

$$
\text{edges}
\rightarrow
\text{shapes}
\rightarrow
\text{object parts}
\rightarrow
\text{object identity}.
$$

---

### 4. Variability

A real object may appear with different:

- positions;
- scales;
- lighting;
- backgrounds;
- orientations;
- occlusions.

Question 1 does not contain this complexity.

---

### 5. Model complexity

A single perceptron is sufficient for Question 1.

Real-world image recognition often uses multilayer or convolutional neural networks to learn complex nonlinear representations.

---

## Final conclusion

$$
\boxed{
\text{Question 1 is a simplified prototype of supervised recognition, not a full equivalent of real-world image recognition.}
}

# MENG 417 Homework #1 — Mature Cheat Sheet

## 1. Linear decision boundary

A perceptron uses

$$
a
=
\mathbf{w}^T\mathbf{x}+b.
$$

The decision boundary is where

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0.
}
$$

In two dimensions, this is a line.

In three dimensions, this is a plane.

In higher dimensions, it is called a hyperplane.

---

## 2. Weight vector

For

$$
w_1x_1+w_2x_2+\cdots+w_nx_n+b=0,
$$

the ordinary weight vector is

$$
\boxed{
\mathbf{w}
=
[w_1,w_2,\ldots,w_n]^T.
}
$$

The bias is

$$
\boxed{b}.
$$

With an augmented input ending in 1, we can combine the bias with the weights.

---

## 3. Geometric meaning of $\mathbf{w}$

The vector

$$
\mathbf{w}
$$

is normal — perpendicular — to the decision hyperplane.

For

$$
2x_1+3x_2+4x_3-4=0,
$$

the normal vector is

$$
\boxed{[2,3,4]^T}.
$$

---

## 4. Binary perceptron classifier

A common $-1/+1$ convention is

$$
\hat y
=
\operatorname{sign}
(
\mathbf{w}^T\mathbf{x}+b
).
$$

A common $0/1$ step convention is

$$
\hat y
=
\begin{cases}
1,&\mathbf{w}^T\mathbf{x}+b\ge0,\\
0,&\mathbf{w}^T\mathbf{x}+b<0.
\end{cases}
$$

Always state which convention you use.

---

## 5. Perceptron learning rule

If a point is misclassified,

$$
\boxed{
\mathbf{w}
\leftarrow
\mathbf{w}
+
\eta y\mathbf{x}
}
$$

and

$$
\boxed{
b
\leftarrow
b+\eta y.
}
$$

Here,

$$
\eta>0
$$

is the learning rate.

---

## 6. Correct-classification test

With labels

$$
y\in\{-1,+1\},
$$

a point is correctly classified when

$$
\boxed{
y(\mathbf{w}^T\mathbf{x}+b)>0.
}
$$

If the quantity is zero, the point lies exactly on the decision boundary.

---

## 7. Perceptron convergence

If the training data are linearly separable, the classical perceptron algorithm will converge to **a** separating hyperplane.

Important:

$$
\boxed{
\text{The separating solution does not have to be unique.}
}
$$

The final weights can depend on:

- initial weights;
- data order;
- learning rate;
- label convention.

What must be true is that the learned boundary separates the classes.

---

## 8. Bias term

The bias shifts the decision boundary.

Without bias,

$$
\mathbf{w}^T\mathbf{x}=0
$$

must pass through the origin.

With bias,

$$
\mathbf{w}^T\mathbf{x}+b=0
$$

can move away from the origin.

---

## 9. Perceptron block diagram

The basic structure is

$$
\boxed{
\text{inputs}
\rightarrow
\text{weights}
\rightarrow
\text{sum + bias}
\rightarrow
\text{activation}
\rightarrow
\text{class}
}
$$

Mathematically,

$$
a
=
\sum_iw_ix_i+b.
$$

Then,

$$
\hat y=\phi(a).
$$

---

## 10. Linear vs nonlinear classification

A single perceptron can only create a linear boundary.

If the classes require a nonlinear boundary, one perceptron is not sufficient.

A multilayer neural network introduces nonlinear transformations that can represent much more complex decision functions.

---

## 11. Supervised learning

Supervised learning uses labeled examples:

$$
(\mathbf{x}_i,y_i).
$$

The objective is to learn a mapping

$$
f(\mathbf{x})
\approx y.
$$

Question 1(b) is supervised learning because the red/blue labels guide the perceptron updates.

---

## 12. Deep learning for classification

Deep learning is especially useful when:

- input dimensionality is high;
- features must be learned automatically;
- class boundaries are nonlinear;
- the dataset is sufficiently large;
- the task contains hierarchical structure.

Examples include:

- image recognition;
- speech recognition;
- natural-language classification;
- medical-image analysis;
- complex cybersecurity classification.

---

## 13. Image recognition connection

A simple perceptron problem and an image-recognition problem both have the abstract structure

$$
\boxed{
\text{numerical input}
\rightarrow
\text{learned model}
\rightarrow
\text{class}.
}
$$

However, image recognition generally involves:

- many more features;
- nonlinear relationships;
- spatial patterns;
- more training samples;
- deeper networks.

# Common Mistakes to Avoid

### Mistake 1 — Forgetting the bias

Do not write only

$$
\mathbf{w}^T\mathbf{x}.
$$

The perceptron decision is usually

$$
\mathbf{w}^T\mathbf{x}+b.
$$

### Mistake 2 — Mixing the bias into $\mathbf{w}$ without explaining it

Ordinary form:

$$
\mathbf{w}=[2,3,4]^T,
\qquad
b=-4.
$$

Augmented form:

$$
\tilde{\mathbf{w}}=[2,3,4,-4]^T.
$$

Both are valid when clearly defined.

### Mistake 3 — Assuming the perceptron has only one possible final line

A linearly separable problem often has many valid separators.

### Mistake 4 — Changing the perceptron weights for correctly classified points

The classical update is applied only when a point is misclassified or, under the convention used here, lies on the boundary.

### Mistake 5 — Calling Question 1(a) algorithmic learning

Question 1(a) is solved manually by inspection.

Question 1(b) is the actual learning algorithm.

### Mistake 6 — Saying the toy problem is fully equivalent to image recognition

It illustrates the same classification principle, but real-world images are much higher-dimensional and more complex.

$$
\boxed{
\mathbf{x}
\rightarrow
\mathbf{w}^T\mathbf{x}+b
\rightarrow
\text{activation}
\rightarrow

}
$$

Training means adjusting

and

until the training examples are correctly separated.
